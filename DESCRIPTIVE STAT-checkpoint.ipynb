# 8.Calculate IQR(difference between 75% and 25% quartile

```python
income_df.plot(x="Mthly_HH_Income", y="Mthly_HH_Expense")
Q1 = income_df["Mthly_HH_Expense"].quantile(0.25)
Q3 = income_df["Mthly_HH_Expense"].quantile(0.75)
IQR = Q3 - Q1
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```
